## set

```python
s=set(['u', 'd', 'ud', 'du', 'u', 'ud'])
s
```

```python
t=set(['d', 'u', 'dd', 'uu'])
t
```

```python
s.union(t)
```

```python
s.intersection(t)
```

```python
s.difference(t)
```

```python
t.difference(s)
```

```python
s.symmetric_difference(t)
```
